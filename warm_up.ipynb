Q1° How many games per season?

```python
query = """
    SELECT SEASON_ID, COUNT(*) AS n_games
    FROM game
    GROUP BY SEASON_ID
    ORDER BY SEASON_ID
"""
pd.read_sql(query, con)
```

Q2° Tallest player

```python
query = """
    SELECT 
        PERSON_ID,
        DISPLAY_FIRST_LAST AS player_name,
        HEIGHT
    FROM common_player_info
    ORDER BY HEIGHT DESC
    LIMIT 10
"""
pd.read_sql(query, con)
```

Q3° Which team have the most experienced player?

```python
query = """
SELECT 
    team_name,
    AVG(season_exp) AS avg_experience,
    COUNT(*) AS num_players
FROM common_player_info
WHERE team_name IS NOT NULL
GROUP BY team_name
ORDER BY avg_experience DESC
LIMIT 10;
"""

pd.read_sql(query, con)
```

Q4° Heaviest player by position

```python
query = """
SELECT 
    position,
    display_first_last AS player_name,
    weight,
    height
FROM common_player_info
WHERE position IS NOT NULL AND weight IS NOT NULL
ORDER BY weight DESC
LIMIT 10;
"""
pd.read_sql(query, con)
```
